```python
# Proyecto: Análisis financiero (Ingeniería de Software)
# Tasa efectiva mensual
tasa = 0.01

def vp_flujo(monto, mes, tasa=tasa):
    """Valor presente de un monto recibido en el mes 'mes'"""
    return monto / ((1 + tasa) ** mes)

def vp_anualidad(pago, meses, tasa=tasa):
    """Valor presente de pagos periódicos al final de cada mes"""
    if tasa == 0:
        return pago * meses
    return pago * (1 - (1 + tasa) ** (-meses)) / tasa

def vpn_proyecto(costo_mensual, meses, beneficio_final, tasa=tasa):
    """Devuelve (VPN, VP_inversión) de un proyecto"""
    vp_inversion = vp_anualidad(costo_mensual, meses, tasa)  # negativo
    vp_beneficio = vp_flujo(beneficio_final, meses, tasa)
    return vp_inversion + vp_beneficio, vp_inversion

# Datos base
costo_mensual = -1000.0
beneficio_final = 18000.0

# (a) Proyecto en 12 meses
vpn_12, vp_inv_12 = vpn_proyecto(costo_mensual, 12, beneficio_final, tasa)

# (b) Proyecto en 15 meses
vpn_15, vp_inv_15 = vpn_proyecto(costo_mensual, 15, beneficio_final, tasa)

# (c) Rentabilidad = VPN / |VP inversión|
rentabilidad_12 = vpn_12 / abs(vp_inv_12)
rentabilidad_15 = vpn_15 / abs(vp_inv_15)

# (d) Proyecto en 18 meses
vpn_18, vp_inv_18 = vpn_proyecto(costo_mensual, 18, beneficio_final, tasa)
perdida_vs_12 = vpn_12 - vpn_18

# (e) Gestión profesional (+5% mensual de costo, 12 meses)
costo_mensual_pm = costo_mensual * 1.05
vpn_12_pm, vp_inv_12_pm = vpn_proyecto(costo_mensual_pm, 12, beneficio_final, tasa)
rentabilidad_12_pm = vpn_12_pm / abs(vp_inv_12_pm)
mejora_pm_vs_15 = vpn_12_pm - vpn_15
mejora_pm_vs_18 = vpn_12_pm - vpn_18

# (f) Desafío: costo mensual máximo para 15 meses y VPN igual al del plan original
vp_beneficio_15 = vp_flujo(beneficio_final, 15, tasa)
costo_max_15 = ((vpn_12 - vp_beneficio_15) * tasa) / (1 - (1 + tasa) ** (-15))

def dinero(x):
    return f"${x:,.2f}"

print("(a) VPN 12 meses:", dinero(vpn_12))
print("(a) VP inversión 12 meses:", dinero(vp_inv_12))
print("(b) VPN 15 meses:", dinero(vpn_15))
print("(b) VP inversión 15 meses:", dinero(vp_inv_15))
print("(c) Rentabilidad 12 meses:", f"{100*rentabilidad_12:.2f}%")
print("(c) Rentabilidad 15 meses:", f"{100*rentabilidad_15:.2f}%")
print("(d) VPN 18 meses:", dinero(vpn_18))
print("(d) Pérdida vs 12 meses:", dinero(perdida_vs_12))
print("(e) VPN con PM 12 meses:", dinero(vpn_12_pm))
print("(e) Rentabilidad con PM:", f"{100*rentabilidad_12_pm:.2f}%")
print("(e) Mejora NPV PM vs 15 meses:", dinero(mejora_pm_vs_15))
print("(e) Mejora NPV PM vs 18 meses:", dinero(mejora_pm_vs_18))
print("(f) Máx. costo mensual aceptable en 15 meses (igualar VPN de (a)):", dinero(costo_max_15))
```
